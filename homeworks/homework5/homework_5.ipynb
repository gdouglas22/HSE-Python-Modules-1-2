## Задание 1

```python
from datetime import datetime

moscow_times = "Wednesday, October 2, 2002"
guardian = "Friday, 11.10.13"
daily_news = "Thursday, 18 August 1977"

fmt_moscow_times = "%A, %B %d, %Y"   
fmt_guardian = "%A, %d.%m.%y"        
fmt_daily_news = "%A, %d %B %Y"      

dt_moscow_times = datetime.strptime(moscow_times, fmt_moscow_times)
dt_guardian = datetime.strptime(guardian, fmt_guardian)
dt_daily_news = datetime.strptime(daily_news, fmt_daily_news)

print(dt_moscow_times)
print(dt_guardian)
print(dt_daily_news)
```
