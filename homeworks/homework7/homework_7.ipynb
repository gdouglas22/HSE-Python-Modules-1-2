## Задание 1

```python
class Client:
    def __init__(self, full_name, gender, age, device, browser, amount, region):
        self.full_name = full_name
        self.gender = (gender or "").strip().lower()
        self.age = int(age)
        self.device = (device or "").strip().lower()
        self.browser = browser
        self.amount = amount
        self.region = region

    def _gender_phrase(self):
        if self.gender in ("female", "женский", "женщина"):
            return "женского пола"
        elif self.gender in ("male", "мужской", "мужчина"):
            return "мужского пола"
        else:
            return "неопределённого пола"

    def _purchase_verb(self):
        if self.gender in ("female", "женский", "женщина"):
            return "совершила"
        elif self.gender in ("male", "мужской", "мужчина"):
            return "совершил"
        return "совершил покупку"

    def _age_word(self):
        n = self.age
        last_two = n % 100
        last = n % 10
        if 11 <= last_two <= 14:
            return "лет"
        if last == 1:
            return "год"
        if 2 <= last <= 4:
            return "года"
        return "лет"

    def _device_phrase(self):
        if self.device == "mobile":
            return "мобильного браузера"
        if self.device == "desktop":
            return "настольного браузера"
        if self.device == "tablet":
            return "планшетного браузера"
        return "браузера"

    def to_description(self):
        gender_phrase = self._gender_phrase()
        verb = self._purchase_verb()
        age_word = self._age_word()
        device_phrase = self._device_phrase()
        text = (
            "Пользователь {name} {gender_phrase}, {age} {age_word} {verb} покупку "
            "на {amount} у.е. с {device_phrase} {browser}. "
            "Регион, из которого совершалась покупка: {region}."
        )

        return text.format(
            name=self.full_name,
            gender_phrase=gender_phrase,
            age=self.age,
            age_word=age_word,
            verb=verb,
            amount=self.amount,
            device_phrase=device_phrase,
            browser=self.browser,
            region=self.region
        )


def parse_client_block(lines):
    data = {}
    for line in lines:
        if ":" in line:
            key, value = line.split(":", 1)
            data[key.strip()] = value.strip()

    full_name = data.get("ФИО", "")
    gender = data.get("Пол", "")
    age = data.get("Возраст", "0")
    device = data.get("Устройство, с которого выполнялась покупка", "")
    browser = data.get("Браузер", "")
    amount = data.get("Сумма чека", "0")
    region = data.get("Регион покупки", "")

    return Client(full_name, gender, age, device, browser, amount, region)


def read_clients_from_file(input_filename):
    clients = []
    with open(input_filename, "r", encoding="utf-8") as f:
        block = []
        for raw_line in f:
            line = raw_line.strip()
            if line == "":
                if block:
                    client = parse_client_block(block)
                    clients.append(client)
                    block = []
            else:
                block.append(line)
        if block:
            client = parse_client_block(block)
            clients.append(client)

    return clients


def write_descriptions_to_file(output_filename, descriptions):
    with open(output_filename, "w", encoding="utf-8") as f:
        for desc in descriptions:
            f.write(desc + "\n")


def generate_client_descriptions(input_filename, output_filename):
    clients = read_clients_from_file(input_filename)
    descriptions = []

    for client in clients:
        desc = client.to_description()
        descriptions.append(desc)

    write_descriptions_to_file(output_filename, descriptions)


if __name__ == "__main__":
    input_file = "clients.csv"
    output_file = "descriptions.txt"

    generate_client_descriptions(input_file, output_file)
    print("Описания записаны в файл:", output_file)
```
