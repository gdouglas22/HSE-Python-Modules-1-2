## Задание 1

```python
import csv

purchases = {}
with open('purchase_log.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        purchases[row['user_id']] = row['category']

with open('visit_log.csv', 'r', encoding='utf-8') as f_in, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as f_out:

    reader = csv.DictReader(f_in)
    fieldnames = ['user_id', 'source', 'category']
    writer = csv.DictWriter(f_out, fieldnames=fieldnames)
    writer.writeheader()

    for row in reader:
        user_id = row['user_id']
        if user_id in purchases:
            writer.writerow({
                'user_id': user_id,
                'source': row['source'],
                'category': purchases[user_id]
            })
```
