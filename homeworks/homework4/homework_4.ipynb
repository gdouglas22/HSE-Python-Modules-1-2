## Задание 1

```python
documents = [
    {'type': 'passport',  'number': '2207 876234', 'name': 'Василий Гупкин'},
    {'type': 'invoice',   'number': '11-2',        'name': 'Геннадий Покемонов'},
    {'type': 'insurance', 'number': '10006',       'name': 'Аристарх Павлов'}
]

directories = {
    '1': ['2207 876234', '11-2'],
    '2': ['10006'],
    '3': []
}

def find_owner_by_doc_number(documents_list, doc_number):
    for doc in documents_list:
        if doc.get('number') == doc_number:
            return doc.get('name')
    return None

def cmd_print_owner(documents_list):
    doc_number = input("\nВведите номер документа:\n").strip()
    owner = find_owner_by_doc_number(documents_list, doc_number)
    if owner:
        print(f"\nРезультат:\nВладелец документа: {owner}")
    else:
        print("\nРезультат:\nДокумент с таким номером не найден.")

def run_cli(documents_list, directories_dict):
    commands = {
        'p': lambda: cmd_print_owner(documents_list),
    }

    while True:
        cmd = input("\nВведите команду (p — владелец по номеру, q — выход):\n").strip().lower()
        if cmd == 'q':
            print("Выход из программы. До свидания!")
            break
        handler = commands.get(cmd)
        if handler:
            handler()
        else:
            print("Неизвестная команда. Попробуйте ещё раз.")

run_cli(documents, directories)
```
